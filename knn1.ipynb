```python
import pandas as pd
import math
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score,classification_report

df=pd.read_csv("data.csv")
print(df.info())
print(f"Duplicates :{df.duplicated().sum()}")
df.drop(['id','Unnamed: 32'],axis=True,inplace=True)

# converting diagnosis feature from categorical to numerical
df['diagnosis'] = df['diagnosis'].map({'B':0,'M':1})

#independent variable
target=df["diagnosis"]
features=df.drop(["diagnosis"],axis=1,inplace=True)
# features=df[['radius_mean', 'texture_mean', 'perimeter_mean',
#        'area_mean', 'smoothness_mean', 'compactness_mean', 'concavity_mean',
#        'concave points_mean', 'symmetry_mean', 'fractal_dimension_mean',
#        'radius_se', 'texture_se', 'perimeter_se', 'area_se', 'smoothness_se',
#        'compactness_se', 'concavity_se', 'concave points_se', 'symmetry_se',
#        'fractal_dimension_se', 'radius_worst', 'texture_worst',
#        'perimeter_worst', 'area_worst', 'smoothness_worst',
#        'compactness_worst', 'concavity_worst', 'concave points_worst',
#        'symmetry_worst', 'fractal_dimension_worst']]
# for i in df.columns:
#     if i!="diagnosis":
#         features.append(i)

print(features)
print(target)
print((math.sqrt(569)))
target
```

```python
x_train,x_test,y_train,y_test=train_test_split(features,target,test_size=0.2,random_state=42)
scaler=StandardScaler()
x_train_scaled=scaler.fit_transform(x_train)
x_test_scaled=scaler.transform(x_test)

knn=KNeighborsClassifier(n_neighbors=int(math.sqrt(569)),metric='euclidean')
knn.fit(x_train_scaled,y_train)
pridiction=knn.predict(x_test_scaled)

print(f"The Acuracy of the system is :{accuracy_score(y_test,pridiction)*100:.0f} %")

new_patient = pd.DataFrame({
    "radius_mean":[12.5],
    "texture_mean":[15.8],
    "perimeter_mean":[80.2],
    "area_mean":[480],
    "smoothness_mean":[0.09],
    "compactness_mean":[0.06],
    "concavity_mean":[0.03],
    "concave points_mean":[0.02],
    "symmetry_mean":[0.18],
    "fractal_dimension_mean":[0.06],
    "radius_se":[0.30],
    "texture_se":[1.00],
    "perimeter_se":[2.10],
    "area_se":[24],
    "smoothness_se":[0.006],
    "compactness_se":[0.015],
    "concavity_se":[0.020],
    "concave points_se":[0.008],
    "symmetry_se":[0.020],
    "fractal_dimension_se":[0.003],
    "radius_worst":[14.0],
    "texture_worst":[20.0],
    "perimeter_worst":[90],
    "area_worst":[600],
    "smoothness_worst":[0.12],
    "compactness_worst":[0.15],
    "concavity_worst":[0.10],
    "concave points_worst":[0.08],
    "symmetry_worst":[0.28], 
    "fractal_dimension_worst":[0.08]
})
new_patient_scaled=scaler.transform(new_patient)
prediction=knn.predict(new_patient_scaled)
if prediction[0]=='M'or 'm':
    print("Result is Postive :\nThe patient has diagnose with breast cancer")
else:
    print("\nResult is Negative:\nThe patient has no symptoms of the breast cancer resut")
    

```
